```python
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
```

```python
dataset
```

```python
dataset.columns
```

```python
dataset=pd.get_dummies(dataset,drop_first=True,dtype=int)
```

```python
dataset
```

```python
dataset.columns
```

```python
independent=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
```

```python
independent
```

```python
dependent=dataset[["charges"]]
```

```python
dependent
```

```python
#Gridsearch and Model Creation:
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
param_grid={'kernel':['linear','poly','rbf','sigmoid'],'C':[10,100,1000,2000,3000]}

grid=GridSearchCV(SVR(),param_grid,refit=True,verbose=3,n_jobs=-1)
grid.fit(independent,dependent)
```

```python
#print best parameter after tuning:
#print grid.best_params(best parameters from grid):
re=grid.cv_results_

print("The R_score value for best parameter{}:".format(grid.best_params_))
```

```python
re
```

```python
table=pd.DataFrame.from_dict(re)
```

```python
table
```

```python
age_input=float(input("Enter the age:"))
bmi_input=float(input("Enter the bmi:"))
sex_male_input=int(input("Sex Male 0 or 1:"))
children_input=int(input("Enter the children:"))
smoker_yes_input=int(input("smoker yes 0 or 1"))
```

```python
Future_prediction=grid.predict([[age_input,bmi_input,sex_male_input,children_input,smoker_yes_input]])# change the parameters
print("Future prediction={}".format(Future_prediction))
```
